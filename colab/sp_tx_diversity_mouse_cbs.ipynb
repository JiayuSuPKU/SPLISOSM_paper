## **Visualize spatial transcript diversity in adult mouse brain**

This interactive notebook allows you to plot spatially variably processed genes detected in adult mouse coronal brain sections (CBS) by three spatial transcriptomics (ST) platforms:
* SR (10X Visium fresh frozen + short-read): CBS
* SiT (10X Visium + Oxford Nanopore): CBS1, CBS2
* Xenium (10X Xenium Prime 5K, predesigned): CBS

#### **1.2 Define data loading functions**

````python
### Load Xenium Prime 5K mouse CBS data
def load_xenium_data(data_dir):
    xenium_dir = f"{data_dir}/mouse_xenium_5k/cbs/"

    # Load the per-codeword density maps
    pickle_file_path = f"{xenium_dir}/all_multi_codeword_genes.pkl"
    with open(pickle_file_path, "rb") as f:
        cw_density_dict = pickle.load(f)

    # Process the data to map genes to their corresponding codewords
    gene_cwid_dict = {}
    if cw_density_dict:
        codeword_lists = list(cw_density_dict.keys())
        for codeword in codeword_lists:
            # Assumes format 'GeneName-CodewordID'
            gene = codeword.split('-')[0]
            if gene not in gene_cwid_dict:
                gene_cwid_dict[gene] = []
            gene_cwid_dict[gene].append(codeword)

    # Filter for genes that have more than one codeword, as these are the
    # only ones for which a log ratio plot is meaningful.
    genes_with_multiple_codewords = sorted([
        gene for gene, codewords in gene_cwid_dict.items() if len(codewords) > 1
    ])
    cwid_list = [gene_cwid_dict[g] for g in genes_with_multiple_codewords]
    cwid_list = list(itertools.chain(*cwid_list))
    cw_density_dict_new = {
        k: v for k, v in cw_density_dict.items() if k in cwid_list
    }

    print(f"Loaded Xenium density maps for {len(cw_density_dict_new)} codewords of {len(genes_with_multiple_codewords)} genes.")

    # Load the Xenium probe BED12 file
    bed_xenium = pd.read_csv(
        f"{xenium_dir}/xenium_mouse_5K_gene_expression_panel_probe_locations.bed",
        sep='\t',
        skiprows=1,
        header=None,
        names=['chrom', 'chromStart', 'chromEnd', 'name', 'score', 'strand',
               'thickStart', 'thickEnd', 'itemRgb', 'blockCount', 'blockSizes',
               'blockStarts', 'description'],
        index_col=False
    )

    # Load the Xenium SV test results
    df_sv_pval = pd.read_csv(f"{xenium_dir}/sv_results.csv", index_col=0)

    return {
        'xenium_cw_density_dict': cw_density_dict_new,
        'xenium_gene_cwid_dict': gene_cwid_dict,
        'xenium_bed': bed_xenium,
        'xenium_sv_res': df_sv_pval
    }
````

````python
# Simple GTF parser for gene/transcript/exon features
def parse_gtf_attributes(attr_string):
    attrs = {}
    for attr in attr_string.strip(';').split(';'):
        if attr.strip():
            key_value = attr.strip().split(' ', 1)
            if len(key_value) == 2:
                key = key_value[0]
                value = key_value[1].strip('"')
                attrs[key] = value
    return attrs

### Load SiT mouse CBS data
def load_sit_data(data_dir):
    sit_dir = f"{data_dir}/mouse_sit/"

    # Load anndata
    samples_to_load = ['ONT-CBS1', 'ONT-CBS2']
    adata_iso_dir_list = [
        f"{sit_dir}/cbs1/iso.quant.h5ad",
        f"{sit_dir}/cbs2/iso.quant.h5ad"
    ]
    adata_gene_dir_list = [
        f"{sit_dir}/cbs1/gene.quant.h5ad",
        f"{sit_dir}/cbs2/gene.quant.h5ad"
    ]
    df_sv_dir_list = [
        f"{sit_dir}/cbs1/sv_results.csv",
        f"{sit_dir}/cbs2/sv_results.csv"
    ]
    df_du_dir_list = [
        f"{sit_dir}/cbs1/du_results.csv",
        f"{sit_dir}/cbs2/du_results.csv"
    ]

    adata_iso_cache, adata_gene_cache = {}, {}
    df_sv_dict, df_du_dict = {}, {}

    # Loop through the specified samples and load each into memory
    for sample_id, adata_iso_dir, adata_gene_dir, sv_dir, du_dir in zip(
        samples_to_load, adata_iso_dir_list, adata_gene_dir_list,
        df_sv_dir_list, df_du_dir_list
    ):
        # Isoform-level quantification
        adata_iso_cache[sample_id] = sc.read_h5ad(adata_iso_dir)
        # Gene-level quantification
        adata_gene_cache[sample_id] = sc.read_h5ad(adata_gene_dir)
        # Spatial variability test results
        df_sv_dict[sample_id] = pd.read_csv(sv_dir, index_col=0)
        # Differential usage test results
        df_du_dict[sample_id] = pd.read_csv(du_dir, index_col=False)

    print(f"Loaded isoform and gene quantification anndata for {len(samples_to_load)} SiT sample(s).")

    # Load and parse ONT-based variable isoform
    gtf_ont_file=f"{sit_dir}/cbs_svs.txid.gtf"
    gtf_ont_sv = []
    with open(gtf_ont_file, 'r') as f:
        for line in f:
            if line.startswith('#'):
                continue
            fields = line.strip().split('\t')
            if len(fields) >= 9:
                attrs = parse_gtf_attributes(fields[8])
                gtf_ont_sv.append({
                    'chrom': fields[0],
                    'source': fields[1],
                    'feature': fields[2],
                    'start': int(fields[3]),
                    'end': int(fields[4]),
                    'strand': fields[6],
                    'gene_name': attrs.get('gene_name', ''),
                    'transcript_id': attrs.get('transcript_id', ''),
                    'transcript_name': attrs.get('transcript_name', ''),
                    'gene_id': attrs.get('gene_id', '')
                })

    gtf_ont_sv = pd.DataFrame(gtf_ont_sv)

    return {
        'sit_anndata_iso_dict': adata_iso_cache,
        'sit_anndata_gene_dict': adata_gene_cache,
        'sit_gtf': gtf_ont_sv,
        'sit_sv_res_dict': df_sv_dict,
        'sit_du_res_dict': df_du_dict
    }
````

````python
### Load Visium-SR mouse CBS data
def load_sr_data(data_dir):
    sr_dir = f"{data_dir}/mouse_visium/cbs/"

    # Load anndata and SV results
    samples_to_load = ['SR-CBS']
    adata_iso_dir_list = [f"{sr_dir}/iso.quant.h5ad"]
    adata_gene_dir_list = [f"{sr_dir}/gene.quant.h5ad"]
    df_sv_dir_list = [f"{sr_dir}/sv_results.csv"]
    df_du_dir_list = [f"{sr_dir}/du_results.csv"]

    adata_iso_cache, adata_gene_cache = {}, {}
    df_sv_dict, df_du_dict = {}, {}

    # Loop through the specified samples and load each into memory
    for sample_id, adata_iso_dir, adata_gene_dir, sv_dir, du_dir in zip(
        samples_to_load, adata_iso_dir_list, adata_gene_dir_list,
        df_sv_dir_list, df_du_dir_list
    ):
        # Isoform-level quantification
        adata_iso_cache[sample_id] = sc.read_h5ad(adata_iso_dir)
        # Gene-level quantification
        adata_gene_cache[sample_id] = sc.read_h5ad(adata_gene_dir)
        # Spatial variability test results
        df_sv_dict[sample_id] = pd.read_csv(sv_dir, index_col=0)
        # Differential usage test results
        df_du_dict[sample_id] = pd.read_csv(du_dir, index_col=False)

    print(f"Loaded isoform and gene quantification anndata for {len(samples_to_load)} SR sample(s).")

    # Load SR-based variable TREND regions
    bed_sr_sv_peak = pd.read_csv(
        f"{sr_dir}/svs.exon.bed",
        sep='\t',
        header=None,
        names=['chrom', 'chromStart', 'chromEnd', 'name', 'score', 'strand',
               'thickStart', 'thickEnd', 'itemRgb', 'blockCount', 'blockSizes',
               'blockStarts', 'description'],
        index_col=False
    )

    return {
        'sr_anndata_iso_dict': adata_iso_cache,
        'sr_anndata_gene_dict': adata_gene_cache,
        'sr_bed': bed_sr_sv_peak,
        'sr_sv_res_dict': df_sv_dict,
        'sr_du_res_dict': df_du_dict
    }
````

````python
# Load and parse reference GTF
def load_ref_gtf(data_dir):
    reference_gtf_file=f"{data_dir}/subset_ensembl_102.gtf.gz"
    gtf_ref_data = []

    with pysam.TabixFile(reference_gtf_file) as gtf:
        for line in gtf.fetch(parser=pysam.asGTF()):
            attrs = parse_gtf_attributes(line.attributes)
            gene_name = attrs.get('gene_name') or attrs.get('gene_id')
            gtf_ref_data.append({
                # 'chrom': fields[0],
                'chrom': line.contig,
                'source': line.source,
                'feature': line.feature,
                'start': int(line.start),
                'end': int(line.end),
                'strand': line.strand,
                'gene_id': attrs.get('gene_id', ''),
                'gene_name': attrs.get('gene_name', ''),
                'transcript_id': attrs.get('transcript_id', ''),
                'transcript_name': attrs.get('transcript_name', '')
            })

    gtf_ref_data = pd.DataFrame(gtf_ref_data)

    print(f"Loaded reference GTF for {gtf_ref_data['gene_name'].nunique()} genes.")

    return gtf_ref_data
````

#### **1.4 Define visualization functions**

We will visualize the data using the following widgets:
1. `create_dataframe_widget` to visualize spatial variability and differential usage test results as interactive tables.
2. `create_spatial_isoform_widget` to visualize the isoform-level anndata from the ONT and SR datasets.
3. `create_xenium_density_widget` to visualize Xenium codeword counts per 20µm-x-20µm Xenium bin.
4. `create_tx_structure_widget` to visualize transcript structure of a selected gene.
5. `create_spatial_rbp_widget` to visualize RBP-gene spatial association for ONT and SR samples.

````python
def create_dataframe_widget(df_dict, title):
    """
    Creates a widget with dropdowns to select and display dataframes interactively.
    """
    # Create dropdown for selecting the dataframe
    df_dropdown = widgets.Dropdown(
        options=list(df_dict.keys()),
        description='Select DataFrame:',
        disabled=not df_dict,
        style={'description_width': 'initial'}
    )

    output = widgets.Output()

    def display_selected_df(change):
        """Displays the selected dataframe using data_table."""
        output.clear_output()
        with output:
            selected_key = change['new']
            if selected_key in df_dict:
                display(data_table.DataTable(df_dict[selected_key], include_index=True, num_rows_per_page=10))
            else:
                print("DataFrame not found.")

    # Observe changes in the dropdown and display the selected dataframe
    df_dropdown.observe(display_selected_df, names='value')

    # Create the widget layout
    widget_layout = widgets.VBox([
        widgets.HTML(f"<h3>{title}</h3>"),
        df_dropdown,
        output
    ])

    display(widget_layout)

    # Display the first dataframe by default if available
    if df_dict:
        display_selected_df({'new': df_dropdown.value})
````

````python
# Visualize anndata isoform expression
def create_spatial_isoform_widget(samples_to_load, adata_cache, df_sv_cache):
    """
    Creates a widget to visualize spatial isoform expression.
    """

    # The core plotting logic is kept separate for clarity.
    def plot_spatial_isoforms(sample, gene, layer, adata, df_sv_pval):
        """
        The main plotting function. It now receives the AnnData object directly.
        """
        # --- Data Extraction ---
        isoforms_to_plot = adata.var.index[adata.var['gene_symbol'] == gene]
        sv_res = df_sv_pval.loc[gene, :]

        # Use different title schemes based on sample type
        if sample.startswith('ONT'):
            names_to_plot = adata.var.loc[adata.var['gene_symbol'] == gene, 'transcript_name']
        else:
            names_to_plot = isoforms_to_plot

        if isoforms_to_plot.empty:
            print(f"Gene '{gene}' has no associated isoforms in this sample.")
            return

        # --- Plotting ---
        try:
            with plt.rc_context({"figure.figsize": (3, 3)}):
                sc.pl.spatial(
                    adata,
                    color=list(isoforms_to_plot),
                    title=list(names_to_plot),
                    img_key='lowres',
                    alpha_img=0.3,
                    layer=layer,
                    ncols=len(isoforms_to_plot),
                    show=False  # Prevent double plotting
                )
                plt.show()  # Explicitly show the plot in the output widget
        except KeyError:
             print(f"Error: The layer '{layer}' does not exist in the AnnData object for sample '{sample}'.")
        except Exception as e:
            print(f"An error occurred during plotting: {e}")

        # Display summary
        print(f"\nSummary for {gene}:")
        print(f"- Avg total count per spot: {sv_res['count_avg']}")
        print(f"- Effective number of isoforms: {sv_res['perplexity']}")
        print(f"- HSIC-IR adjusted p-value: {sv_res['padj_hsic-ir']}")
        print(f"- HSIC-GC adjusted p-value: {sv_res['padj_hsic-gc']}")

    # --- 1. Create Widgets ---
    sample_dropdown = widgets.Dropdown(
        value='ONT-CBS2',
        options=samples_to_load,
        description='Sample:',
        style={'description_width': 'initial'},
        disabled=not samples_to_load
    )

    gene_input = widgets.Combobox(
        value='Dtnbp1',
        placeholder='Type a gene name...',
        options=[],  # Will be populated dynamically
        description='Gene:',
        style={'description_width': 'initial'},
        disabled=True # Initially disabled until genes are loaded
    )

    layer_dropdown = widgets.Dropdown(
        options=['counts', 'log1p', 'ratios_obs', 'ratios_smoothed'],
        value='log1p',
        description='Layer:',
        style={'description_width': 'initial'},
        disabled=not samples_to_load
    )

    visualize_button = widgets.Button(
        description='Visualize',
        button_style='primary'
    )

    output = widgets.Output()

    # --- 2. Define Handler and Observer Functions ---

    def update_gene_options(change):
        """Observer function to update the gene list when the sample changes."""
        with output:
            sample = change['new']
            adata = adata_cache.get(sample)
            if adata is not None:
                # Get unique, sorted gene symbols
                unique_genes = sorted(adata.var['gene_symbol'].dropna().unique())
                gene_input.options = unique_genes
                gene_input.disabled = False
                # If the current gene isn't in the new list, reset to the first available gene
                if gene_input.value not in unique_genes:
                    gene_input.value = unique_genes[0] if unique_genes else ''
            else:
                gene_input.options = []
                gene_input.disabled = True
                clear_output()
                print(f"Data for sample '{sample}' not loaded. Cannot populate gene list.")

    def visualize_handler(b):
        """The main handler function called when the 'Visualize' button is clicked."""
        output.clear_output()

        with output:
            # --- Input Validation ---
            sample = sample_dropdown.value
            gene = gene_input.value.strip()
            layer = layer_dropdown.value

            if not sample:
                print("Please select a sample.")
                return
            if not gene:
                print("Please type or select a gene name.")
                return

            adata = adata_cache.get(sample)
            df_sv_pval = df_sv_cache.get(sample)
            if adata is None:
                print(f"Error: Data for sample '{sample}' is not loaded.")
                return

            if gene not in adata.var['gene_symbol'].values:
                print(f"Gene '{gene}' not found in sample '{sample}'. Please check the spelling.")
                return

            print(f"Generating plot for gene '{gene}' in sample '{sample}' using layer '{layer}'...")

            # Call the core plotting function
            plot_spatial_isoforms(sample, gene, layer, adata, df_sv_pval)

    # --- 3. Connect Widgets to Functions ---
    sample_dropdown.observe(update_gene_options, names='value')
    visualize_button.on_click(visualize_handler)

    # --- 4. Assemble and Display the Layout ---
    controls = widgets.HBox([sample_dropdown, gene_input, layer_dropdown, visualize_button])
    ui = widgets.VBox([
        widgets.HTML("<h3>Spatial Isoform Expression</h3>"),
        controls,
        output
    ])

    display(ui)

    # --- 5. Initial Call ---
    # Manually trigger the functions once to populate the initial state
    if samples_to_load:
        update_gene_options({'new': sample_dropdown.value})
        visualize_handler(None)
````

````python
# Visualize Xenium per-gene codeword expression
def visualize_gene_codewords(
    gene,
    density_maps,
    gene_cwid_dict,
    scale_factor = 20.0,
    vmax_pct = None,
    gamma = 0.95
):
    """Visualize two codewords and RGB composite for a gene with aligned axes."""

    codeword_list = gene_cwid_dict.get(gene, [])
    if len(codeword_list) < 2:
        raise ValueError(f"Gene '{gene}' needs at least 2 codewords")

    # Extract density maps
    density_map_list = [density_maps[cw].toarray() for cw in codeword_list]
    n_codewords = len(density_map_list)

    # Create figure with GridSpec for better control
    fig = plt.figure(figsize=(6 * (n_codewords + 1), 6))
    gs = gridspec.GridSpec(1, (n_codewords + 1), figure=fig, wspace=0.4)

    axes = [fig.add_subplot(gs[i]) for i in range(n_codewords + 1)]

    # Common extent for all panels
    height, width = density_map_list[0].shape
    extent = [0, width * scale_factor, 0, height * scale_factor]

    # Store images for colorbar creation
    images = []

    # First n_codewords panels: Codeword density heatmaps
    for i, (ax, dense_map, codeword) in enumerate(
        zip(axes[:n_codewords], density_map_list, codeword_list[:n_codewords])
    ):
        if vmax_pct is not None:
            vmax = np.percentile(dense_map[dense_map > 0], vmax_pct)
            im = ax.imshow(dense_map, origin='lower', aspect='equal', cmap='viridis', extent=extent, vmax = vmax)
        else:
            im = ax.imshow(dense_map, origin='lower', aspect='equal', cmap='viridis', extent=extent)

        ax.set_title(f"Codeword: {codeword}")
        ax.set_xlabel("X Position (µm)")
        ax.set_ylabel("Y Position (µm)")
        images.append(im)

    # The last panel: RGB composite
    rgb_image = np.zeros((height, width, 3))
    legend_patches = []

    for i, (dense_map, codeword) in enumerate(zip(density_map_list, codeword_list[:3])):
        channel = i
        normalized = dense_map / (np.max(dense_map) + 1e-9)
        rgb_image[:, :, channel] = np.power(normalized, gamma)

        color = ['red', 'green', 'blue'][i]
        legend_patches.append(
            mpatches.Patch(color=color, label=f"{gene}-{codeword.split('-')[-1]}")
        )

    axes[-1].imshow(rgb_image, origin='lower', aspect='equal', extent=extent)
    axes[-1].set_xlabel("X Position (µm)")
    axes[-1].set_ylabel("Y Position (µm)")
    axes[-1].legend(handles=legend_patches, loc='upper center', ncol = len(density_map_list))
    axes[-1].set_title(f"RGB Composite: {gene}")

    # Add colorbars to first panels with consistent positioning
    for i, (ax, im) in enumerate(zip(axes[:n_codewords], images)):
        # Create colorbar with same width as the RGB panel's legend area
        divider = make_axes_locatable(ax)
        cax = divider.append_axes("right", size="5%", pad=0.1)
        plt.colorbar(im, cax=cax, label="Transcript count")

    # Ensure all axes have the same position and size
    # Get the position of the third axis (which has no colorbar)
    pos3 = axes[-1].get_position()

    # Adjust the first two axes to match the width of the third
    for i in range(n_codewords):
        pos = axes[i].get_position()
        axes[i].set_position([pos.x0, pos3.y0, pos3.width, pos3.height])

    plt.show()

# Xenium expression widget
def create_xenium_density_widget(
    gene_cwid_dict,
    cw_density_dict,
    df_sv_pval,
    vmax_pct = None,
    gamma = 0.95
):
    """
    Create a widget with a dropdown and button to visualize gene codeword densities.
    """

    # --- Widget Creation ---

    unique_genes = sorted(list(gene_cwid_dict.keys()))

    gene_input = widgets.Combobox(
        value='Dtnbp1' if 'Dtnbp1' in unique_genes else (unique_genes[0] if unique_genes else ''),
        placeholder='Type a gene name...',
        options=unique_genes,
        description='Gene:',
        style={'description_width': 'initial'},
        disabled=not unique_genes
    )

    visualize_button = widgets.Button(
        description='Visualize',
        button_style='primary'
    )

    output = widgets.Output()

    # Define the click handler that calls the plotting function
    def visualize_gene_densities(b):
        output.clear_output()

        with output:
            gene_name = gene_input.value
            if not gene_name:
                print("Please enter or select a gene name.")
                return

            # Call the main plotting function
            visualize_gene_codewords(
                gene_name,
                density_maps = cw_density_dict,
                gene_cwid_dict = gene_cwid_dict,
                vmax_pct=vmax_pct
            )

            # Extract SV results
            sv_res = df_sv_pval.loc[gene_name]

            # Display summary
            print(f"\nSummary for {gene_name}:")
            print(f"- Protein name: {sv_res['protein_name']}")
            print(f"- Location: {sv_res['location']}")
            print(f"- Cell type: {sv_res['cell_type']}")
            print(f"- HSIC-IR adjusted p-value: {sv_res['padj_hsic-ir']}")
            print(f"- HSIC-GC adjusted p-value: {sv_res['padj_hsic-gc']}")

    # Connect button to the handler function
    visualize_button.on_click(visualize_gene_densities)

    # Assemble the final layout
    widget_layout = widgets.VBox([
        widgets.HTML("<h3>Codeword Density Visualization</h3>"),
        widgets.HBox([gene_input, visualize_button]),
        output
    ])

    # Display the final widget
    display(widget_layout)

    # Trigger the visualization for the default gene upon initial display
    if gene_input.value:
        visualize_gene_densities(None)
````

````python
# Visualize Xenium probe, ONT-SV and SR-SV events
def create_tx_structure_widget(
    xenium_bed,
    selected_region_bed,
    selected_iso_gtf,
    reference_gtf
) -> None:
    """
    Create a widget with static matplotlib visualization of Xenium data.
    """
    import matplotlib.pyplot as plt
    import matplotlib.patches as patches
    from matplotlib.patches import Rectangle

    # Read files
    xenium_df = xenium_bed
    regions_df = selected_region_bed
    selected_gtf_df = selected_iso_gtf
    gtf_df = reference_gtf

    # Extract gene names from Xenium data
    xenium_df['gene_name'] = xenium_df['name'].str.extract(r'^(\S+)\s+\(')
    unique_genes = sorted(xenium_df['gene_name'].dropna().unique())

    # Create widgets
    gene_input = widgets.Combobox(
        value='Dtnbp1' if 'Dtnbp1' in unique_genes else (unique_genes[0] if unique_genes else ''),
        placeholder='Type a gene name...',
        options=unique_genes,
        description='Gene:',
        style={'description_width': 'initial'}
    )

    visualize_button = widgets.Button(
        description='Visualize',
        button_style='primary'
    )

    output = widgets.Output()

    def visualize_gene(b):
        output.clear_output()

        with output:
            gene_name = gene_input.value
            if not gene_name:
                print("Please enter a gene name")
                return

            # Filter data for the selected gene
            gene_xenium = xenium_df[xenium_df['gene_name'] == gene_name]

            if gene_xenium.empty:
                print(f"No Xenium data found for gene: {gene_name}")
                return

            # Also make sure isoforms are on the main assembly
            chrom = gene_xenium.iloc[0]['chrom']
            chrom_wo_prefix = chrom.replace('chr', '')

            # For Septin genes
            if gene_name.startswith('Sept'):
                gene_name = gene_name.replace('Sept', 'Septin')

            # Get gene region from GTF
            gene_gtf = gtf_df[
                (gtf_df['gene_name'] == gene_name) & \
                    ((gtf_df['chrom'] == chrom) | (gtf_df['chrom'] == chrom_wo_prefix))
            ]
            selected_gene_gtf = selected_gtf_df[selected_gtf_df['gene_name'] == gene_name]

            if gene_gtf.empty:
                print(f"No GTF data found for gene: {gene_name}")
                return

            # Determine visualization region
            strand = gene_gtf.iloc[0]['strand']
            all_starts = list(gene_xenium['chromStart']) + list(gene_gtf['start'])
            all_ends = list(gene_xenium['chromEnd']) + list(gene_gtf['end'])
            max_length = abs(min(all_starts) - max(all_ends))
            region_start = min(all_starts) - max_length * 0.1
            region_end = max(all_ends) + max_length * 0.1

            # Create figure with 4 subplots
            fig, (ax1, ax2, ax3, ax4) = plt.subplots(
                4, 1, figsize=(16, 8),
                gridspec_kw={'height_ratios': [3, 2, 1, 1]}
            )

            # Format x-axis labels without scientific notation
            def format_axis(ax):
                ax.ticklabel_format(style='plain', axis='x')
                ax.get_xaxis().get_major_formatter().set_useOffset(False)

                # Turn off y axis labels
                ax.tick_params(axis='y', labelleft=False)

                # Set fontsize to 12
                for item in ([ax.title, ax.xaxis.label, ax.yaxis.label] + ax.get_xticklabels()):
                    item.set_fontsize(12)

            # Plot 1: Reference gene isoforms from GTF
            ax1.set_title(f"Reference Gene Structure: {gene_name}")
            ax1.set_xlim(region_start, region_end)
            format_axis(ax1)

            # Get unique transcripts
            transcripts = gene_gtf[gene_gtf['feature'] == 'transcript']['transcript_id'].unique()

            for i, transcript_id in enumerate(transcripts):
                y_pos = i * 0.5

                # Get transcript data
                transcript_data = gene_gtf[gene_gtf['transcript_id'] == transcript_id]
                transcript_name = transcript_data[transcript_data['feature'] == 'transcript']['transcript_name'].iloc[0]

                # Draw transcript line
                transcript_start = transcript_data[transcript_data['feature'] == 'transcript']['start'].iloc[0]
                transcript_end = transcript_data[transcript_data['feature'] == 'transcript']['end'].iloc[0]
                ax1.plot([transcript_start, transcript_end], [y_pos, y_pos], 'k-', linewidth=1)

                # Draw exons
                exons = transcript_data[transcript_data['feature'] == 'exon']
                for _, exon in exons.iterrows():
                    rect = Rectangle((exon['start'], y_pos - 0.15),
                                   exon['end'] - exon['start'], 0.3,
                                   facecolor='darkblue', edgecolor='black')
                    ax1.add_patch(rect)

                # Add transcript label
                ax1.text(
                    region_start + 0.01*max_length, y_pos, transcript_name,
                    fontsize=12, va='center'
                )

            # Add strand information
            strand_info = "5'>3'" if strand == '+' else "3'<5'"
            ax1.text(region_end - 0.05*max_length, y_pos, strand_info, fontsize=12, va='center')

            ax1.set_ylim(-0.5, len(transcripts) * 0.5)
            ax1.set_ylabel('Reference\ntranscripts')
            ax1.set_xlabel('')

            # Plot 2: Visium-ONT isoforms
            ax2.set_title("Spatially variable isoforms (ONT)")
            ax2.set_xlim(region_start, region_end)
            format_axis(ax2)

            selected_transcripts = selected_gene_gtf[selected_gene_gtf['feature'] == 'transcript']['transcript_id'].unique()

            for i, transcript_id in enumerate(selected_transcripts):
                y_pos = i * 0.5

                # Get transcript data
                transcript_data = selected_gene_gtf[selected_gene_gtf['transcript_id'] == transcript_id]
                transcript_name = transcript_data[transcript_data['feature'] == 'transcript']['transcript_name'].iloc[0]

                # Draw transcript line
                transcript_rows = transcript_data[transcript_data['feature'] == 'transcript']
                if not transcript_rows.empty:
                    transcript_start = transcript_rows['start'].iloc[0]
                    transcript_end = transcript_rows['end'].iloc[0]
                    ax2.plot([transcript_start, transcript_end], [y_pos, y_pos], 'k-', linewidth=1)

                # Draw exons
                exons = transcript_data[transcript_data['feature'] == 'exon']
                for _, exon in exons.iterrows():
                    rect = Rectangle((exon['start'], y_pos - 0.15),
                                   exon['end'] - exon['start'], 0.3,
                                   facecolor='red', edgecolor='black')
                    ax2.add_patch(rect)

                # Add transcript label
                ax2.text(
                    region_start + 0.01*max_length, y_pos, transcript_name,
                    fontsize=12, va='center'
                )

            if len(selected_transcripts) > 0:
                ax2.set_ylim(-0.5, len(selected_transcripts) * 0.5)
            else:
                ax2.set_ylim(-0.5, 0.5)
                ax2.text(
                    region_start + (region_end - region_start) / 2, 0,
                    'No detected variable isoforms', ha='center', va='center',
                    fontsize=12
                )

            ax2.set_ylabel('Selected\nisoforms')
            ax2.set_xlabel('')

            # Plot 3: Visium-SR TREND regions
            ax3.set_title("Spatially variable TREND events (SR)")
            ax3.set_xlim(region_start, region_end)
            format_axis(ax3)

            # Filter regions for this chromosome and range
            filtered_regions = regions_df[
                (regions_df['chrom'] == chrom) &
                (regions_df['chromEnd'] >= region_start) &
                (regions_df['chromStart'] <= region_end)
            ]

            for i, (_, row) in enumerate(filtered_regions.iterrows()):
                y_pos = i * 0.3

                # Draw codeword
                rect = Rectangle((row['chromStart'], y_pos - 0.1),
                               row['chromEnd'] - row['chromStart'], 0.2,
                               facecolor='green', edgecolor='black')
                ax3.add_patch(rect)

                # Add label
                ax3.text(
                    row['chromStart'], y_pos + 0.15, row['name'],
                    fontsize=12, ha='left'
                )

            if len(filtered_regions) == 0:
                ax3.text(
                    region_start + (region_end - region_start) / 2, 0,
                    'No detected variable TREND events', ha='center', va='center',
                    fontsize=12
                )
                ax3.set_ylim(-0.3, 0.3)
            else:
                ax3.set_ylim(-0.3, len(filtered_regions) * 0.3)

            ax3.set_ylabel('Regions')
            ax3.set_xlabel('')

            # Plot 4: Xenium codewords
            ax4.set_title("Xenium Prime 5K codewords")
            ax4.set_xlim(region_start, region_end)
            format_axis(ax4)

            for i, (_, row) in enumerate(gene_xenium.iterrows()):
                y_pos = i * 0.3

                # Parse RGB color
                rgb = [int(x)/255 for x in row['itemRgb'].split(',')]

                # Draw codeword
                rect = Rectangle((row['chromStart'], y_pos - 0.1),
                               row['chromEnd'] - row['chromStart'], 0.2,
                               facecolor=rgb, edgecolor='black')
                ax4.add_patch(rect)

                # Add label
                codeword_info = row['name'].split('|')[1].strip()
                ax4.text(
                    row['chromStart'], y_pos + 0.15, codeword_info,
                    fontsize=12, ha='left'
                )

            ax4.set_ylim(-0.3, len(gene_xenium) * 0.3)
            ax4.set_ylabel('Codewords')
            ax4.set_xlabel('')
            ax4.set_xlabel(f'Genomic position on {chrom}')

            plt.tight_layout()
            plt.show()

            # Display summary
            print(f"\nSummary for {gene_name}:")
            print(f"- Number of reference transcripts: {len(transcripts)}")
            print(f"- Number of overlapping spatially variable isoforms (ONT): {len(selected_transcripts)}")
            print(f"- Number of overlapping spatially variable TREND regions (SR): {len(filtered_regions)}")
            print(f"- Number of Xenium Prime 5K codewords: {len(gene_xenium)}")

    # Connect button
    visualize_button.on_click(visualize_gene)

    # Create layout
    widget_layout = widgets.VBox([
        widgets.HTML("<h3>Transcript Structure Visualization</h3>"),
        widgets.HBox([gene_input, visualize_button]),
        output
    ])

    display(widget_layout)

    # Trigger the visualization for the default gene upon initial display
    if gene_input.value:
        visualize_gene(None)
````

````python
def create_spatial_rbp_widget(samples_to_load, adata_iso_cache, adata_rbp_cache, df_du_cache):
    """
    Creates a widget to visualize spatial RNA binding protein expression and compare with isoform expression
    """

    # The core plotting logic is kept separate for clarity.
    def plot_spatial_rbp_n_isoform(
        sample, df_du_pval,
        gene_iso, layer_iso, adata_iso,
        gene_rbp, layer_rbp, adata_rbp
    ):
        """
        The main plotting function. It now receives the AnnData object directly.
        """
        # --- Data Extraction ---
        du_res = df_du_pval.loc[
            (df_du_pval['gene'] == gene_iso) & (df_du_pval['covariate'] == gene_rbp)
        ]

        if du_res.empty:
            print(f"No association data found for gene {gene_iso} and RBP {gene_rbp} in this sample.")
            return

        isoforms_to_plot = adata_iso.var.index[adata_iso.var['gene_symbol'] == gene_iso]
        rbp_to_plot = adata_rbp.var.index[adata_rbp.var['gene_symbols'] == gene_rbp]

        # Use different title schemes based on sample type
        if sample.startswith('ONT'):
            names_to_plot = adata_iso.var.loc[adata_iso.var['gene_symbol'] == gene_iso, 'transcript_name']
        else:
            names_to_plot = isoforms_to_plot

        if isoforms_to_plot.empty:
            print(f"Gene '{gene_iso}' has no associated isoforms in this sample.")
            return

        if rbp_to_plot.empty:
            print(f"RBP '{gene_rbp}' has no expression data in this sample.")
            return

        # --- Extract expression data for scatter plots ---
        try:
            # Get RBP expression data
            if layer_rbp in adata_rbp.layers:
                rbp_expr = adata_rbp.layers[layer_rbp][:, adata_rbp.var.index.get_loc(rbp_to_plot[0])].toarray().flatten()
            else:
                rbp_expr = adata_rbp.X[:, adata_rbp.var.index.get_loc(rbp_to_plot[0])].toarray().flatten()

            # Get isoform expression data
            isoform_expressions = []
            isoform_names = []

            for i, isoform in enumerate(isoforms_to_plot):
                if layer_iso in adata_iso.layers:
                    try:
                        iso_expr = adata_iso.layers[layer_iso][:, adata_iso.var.index.get_loc(isoform)].toarray().flatten()
                    except:
                        iso_expr = adata_iso.layers[layer_iso][:, adata_iso.var.index.get_loc(isoform)].flatten()
                else:
                    iso_expr = adata_iso.X[:, adata_iso.var.index.get_loc(isoform)].toarray().flatten()

                isoform_expressions.append(iso_expr)

                # Use transcript names for ONT samples, otherwise use isoform IDs
                if sample.startswith('ONT'):
                    name = names_to_plot.iloc[i] if i < len(names_to_plot) else isoform
                else:
                    name = isoform
                isoform_names.append(name)

        except Exception as e:
            print(f"Error extracting expression data: {e}")
            return

        # --- Plotting ---
        n_isoforms = len(isoforms_to_plot)

        # Create subplot layout: 1 spatial plot + n_isoforms scatter plots
        fig = plt.figure(figsize=(4 + 4*n_isoforms, 4))

        # Create gridspec for better control over spacing
        gs = fig.add_gridspec(
            1, n_isoforms + 1, width_ratios=[1] + [1]*n_isoforms,
            wspace=0.5, hspace=0.2
        )

        try:
            # Plot spatial RBP expression
            ax_spatial = fig.add_subplot(gs[0, 0])

            # Use scanpy's spatial plotting but capture the axis
            with plt.rc_context({"figure.figsize": (4, 4)}):
                sc.pl.spatial(
                    adata_rbp,
                    color=rbp_to_plot,
                    img_key='lowres',
                    alpha_img=0.3,
                    layer=layer_rbp,
                    ax=ax_spatial,
                    show=False,
                    frameon=False
                )

            ax_spatial.set_title(f'{gene_rbp} (RBP)\n{layer_rbp}', fontsize=10, pad=10)
            pos_sp = ax_spatial.get_position()

            # Plot scatter plots for each isoform
            for i, (iso_expr, iso_name) in enumerate(zip(isoform_expressions, isoform_names)):
                ax_scatter = fig.add_subplot(gs[0, i + 1])

                # Create scatter plot
                scatter = ax_scatter.scatter(rbp_expr, iso_expr, alpha=0.6, s=20, c='steelblue')

                # Add correlation coefficient
                correlation = np.corrcoef(rbp_expr, iso_expr)[0, 1]

                ax_scatter.set_xlabel(f'{gene_rbp} expression', fontsize=9)
                ax_scatter.set_ylabel(f'Isoform expression', fontsize=9)
                ax_scatter.set_title(f'{iso_name}\nPearson\'s R = {correlation:.2f}', fontsize=10, pad=10)

                # Style the scatter plot
                ax_scatter.grid(True, alpha=0.3)
                ax_scatter.tick_params(axis='both', which='major', labelsize=8)

                # Set equal aspect ratio for better alignment
                ax_scatter.set_aspect('auto')

                # Adjust the axes to align with the first plot
                pos = ax_scatter.get_position()
                ax_scatter.set_position([pos.x0, pos_sp.y0, pos_sp.width, pos_sp.height])

            # Adjust layout
            plt.show()

        except KeyError:
             print(f"Error: The layer '{layer_rbp}' does not exist in the RBP AnnData object for sample '{sample}'.")
        except Exception as e:
            print(f"An error occurred during plotting: {e}")

        # Display summary information
        print(f"\nSummary for {gene_iso} vs {gene_rbp} in sample {sample}:")
        print(f"- Conditional HSIC p-value: {du_res['pvalue_hsic-gp'].values[0]}")
        print(f"- GLMM p-value: {du_res['pvalue_glmm'].values[0]}")
        print(f"- Unconditional HSIC p-value: {du_res['pvalue_hsic'].values[0]}")
        print(f"- GLM p-value: {du_res['pvalue_glm'].values[0]}")

    # --- 1. Create Widgets ---
    sample_dropdown = widgets.Dropdown(
        value='ONT-CBS2',
        options=samples_to_load,
        description='Sample:',
        style={'description_width': 'initial'},
        disabled=not samples_to_load
    )

    gene_input = widgets.Combobox(
        value='Gnas',
        placeholder='Type a gene name...',
        options=[],  # Will be populated dynamically
        description='Gene:',
        style={'description_width': 'initial'},
        disabled=True # Initially disabled until genes are loaded
    )

    rbp_input = widgets.Combobox(
        value='Rbfox3',
        placeholder='Type a RBP name...',
        options=[],  # Will be populated dynamically
        description='RBP:',
        style={'description_width': 'initial'},
        disabled=True # Initially disabled until genes are loaded
    )

    layer_gene_dropdown = widgets.Dropdown(
        options=['counts', 'log1p', 'ratios_obs', 'ratios_smoothed'],
        value='log1p',
        description='Layer (gene):',
        style={'description_width': 'initial'},
        disabled=not samples_to_load
    )

    layer_rbp_dropdown = widgets.Dropdown(
        options=['counts', 'log1p'],
        value='log1p',
        description='Layer (RBP):',
        style={'description_width': 'initial'},
        disabled=not samples_to_load
    )

    visualize_button = widgets.Button(
        description='Visualize',
        button_style='primary'
    )

    output = widgets.Output()

    # --- 2. Define Handler and Observer Functions ---

    def update_gene_options(change):
        """Observer function to update the gene list when the sample changes."""
        with output:
            sample = change['new']
            adata_iso = adata_iso_cache.get(sample)
            adata_rbp = adata_rbp_cache.get(sample)

            if adata_iso is not None:
                # Get unique, sorted gene symbols
                unique_genes = sorted(adata_iso.var['gene_symbol'].dropna().unique())
                gene_input.options = unique_genes
                gene_input.disabled = False
                # If the current gene isn't in the new list, reset to the first available gene
                if gene_input.value not in unique_genes:
                    gene_input.value = unique_genes[0] if unique_genes else ''
            else:
                gene_input.options = []
                gene_input.disabled = True
                clear_output()
                print(f"Data for sample '{sample}' not loaded. Cannot populate gene list.")

            if adata_rbp is not None:
                # Get unique, sorted RBP symbols
                unique_rbps = sorted(adata_rbp.var['gene_symbols'].dropna().unique())
                rbp_input.options = unique_rbps
                rbp_input.disabled = False
                # If the current gene isn't in the new list, reset to the first available gene
                if rbp_input.value not in unique_rbps:
                    rbp_input.value = unique_rbps[0] if unique_rbps else ''
            else:
                rbp_input.options = []
                rbp_input.disabled = True
                clear_output()
                print(f"Data for sample '{sample}' not loaded. Cannot populate RBP list.")

    def visualize_handler(b):
        """The main handler function called when the 'Visualize' button is clicked."""
        output.clear_output()

        with output:
            # --- Input Validation ---
            sample = sample_dropdown.value
            gene = gene_input.value.strip()
            rbp = rbp_input.value.strip()
            layer_gene = layer_gene_dropdown.value
            layer_rbp = layer_rbp_dropdown.value

            if not sample:
                print("Please select a sample.")
                return
            if not gene:
                print("Please type or select a gene name.")
                return
            if not rbp:
                print("Please type or select an RBP name.")
                return

            adata_iso = adata_iso_cache.get(sample)
            adata_rbp = adata_rbp_cache.get(sample)
            df_du_pval = df_du_cache.get(sample)
            if (adata_iso is None) or (adata_rbp is None):
                print(f"Error: Anndata for sample '{sample}' are not loaded.")
                return

            print(f"Generating plot for RBP '{rbp}' (layer='{layer_rbp}') and gene '{gene}' (layer='{layer_gene}') in sample '{sample}'...")

            # Call the core plotting function
            plot_spatial_rbp_n_isoform(
                sample, df_du_pval,
                gene, layer_gene, adata_iso,
                rbp, layer_rbp, adata_rbp
            )

    # --- 3. Connect Widgets to Functions ---
    sample_dropdown.observe(update_gene_options, names='value')
    visualize_button.on_click(visualize_handler)

    # --- 4. Assemble and Display the Layout ---
    controls1 = widgets.HBox([sample_dropdown, gene_input, layer_gene_dropdown])
    controls2 = widgets.HBox([rbp_input, layer_rbp_dropdown, visualize_button])
    ui = widgets.VBox([
        widgets.HTML("<h3>Spatial RBP Expression and Association</h3>"),
        controls1,
        controls2,
        output
    ])

    display(ui)

    # --- 5. Initial Call ---
    # Manually trigger the functions once to populate the initial state
    if samples_to_load:
        update_gene_options({'new': sample_dropdown.value})
        visualize_handler(None)
````
